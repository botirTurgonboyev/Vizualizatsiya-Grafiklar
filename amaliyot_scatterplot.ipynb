```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv("https://github.com/anvarnarz/praktikum_datasets/blob/main/uybor_scrapping.csv?raw=true")
df.head()
```

###1-vazifa

Uylarning narxi bilan size o'rtasidagi bog'liqlik grafik

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x=df['size'], y=df['price'])
plt.show()
```

Maydoni 250dan kattalarini tashlab yuboramiz

```python
df = df[df['size'] < 250]
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x=df['size'], y=df['price'])
plt.show()
```

### 2-VAZIFA. Uylarning narxi (price) va maydoni (size) o'rtasida bog'liqlikni ko'rsatuvchi scatter va to'g'ri chiziqni chizing.

```python
plt.figure(figsize=(10,5))
#sns.scatterplot(data=df, x=df['size'], y=df['price'])
sns.regplot(data=df, x=df['size'], y=df['price'])
plt.show()
```

### 3-VAZIFA. Uylarning narxi, maydoni va xonalar soni o'rtasidagi bog'liqlikni ko'rsating.

```python
df.columns
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='size', y='price', hue='rooms', palette='Set1')
plt.show()
```

### 4-VAZIFA. Uylarning narxi, maydoni va tumanlar o'rtasidagi bog'liqlikni ko'rsating.

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='size', y='price', hue='district', palette='Set1')
plt.show()
```

### 5-VAZIFA. Swarm plot yordamida turli tumanlardag uylarning narxini taqqoslang.

```python
plt.figure(figsize=(10,5))
sns.swarmplot(data=df, x='district', y='price')
plt.xticks(rotation=90)
plt.show()
```
