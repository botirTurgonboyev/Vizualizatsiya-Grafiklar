```python
import numpy as np
import pandas as pd
```

```python
import matplotlib.pyplot as plt
%matplotlib Inline
```

```python
import seaborn as sns
```

```python
df = pd.read_csv("https://github.com/anvarnarz/praktikum_datasets/raw/main/museum_visitors.csv", index_col=0)
df.head()
```

Muzeylarga tashriflar soni haqida database berilgan


```python
df.shape
```

```python
plt.plot(df)
plt.show()
```

```python
sns.lineplot(data=df)
plt.show()
```

```python
plt.figure(figsize=(16, 6))
sns.lineplot(data=df)
plt.xticks(rotation=90)

plt.title('Muzeylarga tashriflar haqida')
plt.xlabel('Yil')
plt.ylabel('Tashriflar soni')

plt.show()
```

```python
plt.figure(figsize=(16, 6))
sns.lineplot(x=df.index, y=df['America Tropical Interpretive Center'])
plt.xticks(rotation=90)

plt.title('America Tropical Interpretive Center')
plt.xlabel('Yil')
plt.ylabel('Tashriflar soni')

plt.show()
```

```python
plt.figure(figsize=(16, 6))
sns.lineplot(data = df[['America Tropical Interpretive Center', 'Avila Adobe']])
plt.xticks(rotation=90)

plt.title('America Tropical Interpretive Center va Avila Adobe')
plt.xlabel('Yil')
plt.ylabel('Tashriflar soni')

plt.show()
```
