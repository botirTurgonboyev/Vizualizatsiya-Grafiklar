```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import pandas as pd
```

Uyboruz saytidan olingan malumotlar bilan ishlaymiz

```python
df = pd.read_csv("https://github.com/anvarnarz/praktikum_datasets/blob/main/uybor_scrapping.csv?raw=true")
df.head()
```

Uylar narxini taqsimot grafikda ko'rsatamiz

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='price')
plt.show()
```

200 k dan qimmat uylarni tashlab yuboramiz

```python
df = df[df.price < 200000]
df.shape
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='price')
plt.show()
```

Maydoni taqsimot grafigi

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='size')
plt.show()
```

```python
#Xonalar soni
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='rooms')
plt.show()
```

xonalari 6 tadan kop uylarni tashlab yuboramiz

```python
df = df[df['rooms'] < 7]
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='rooms')
plt.show()
```

```python
df.shape
```

```python
#Narxini xonalar soniga bogliqlik grafiki
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='price', hue='rooms')
plt.show()
```

1,2,3 xonali uylar uchun chizamiz

```python
sns.displot(data=df[df.rooms < 4], x='price', hue='rooms',multiple='stack', palette='Set2', height=5, aspect=2)
plt.xticks(rotation=90)
plt.grid(which='both')
plt.show()
```

4,5,6 xonali uylarni graafigi alohida grafiklarda chizish

```python
sns.displot(data=df[df['rooms'] > 3], x='price', col='rooms', palette='Set2', height=5, aspect=1)
plt.show()
```

Uylarning maydon taqsimotini KDE bilan chizing

```python
sns.displot(data=df, x='size', kind='kde', height=5, aspect=2)
plt.show()
```

```python
sns.displot(data=df, x='size', kde=True, height=5, aspect=2)
plt.show()
```

2, 3, 5 xonali uylar uchun maydoni bilan kde grafik

```python
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df[df.rooms.isin([2,3,5])], x='size', hue='rooms', shade=True)
plt.show()
```

uylarning maydoni uchun kumulyativ taqsimot grafik

```python
sns.displot(data=df, x='size', kind='ecdf', height=5, aspect=2)
plt.grid(which='both')
plt.minorticks_on()
plt.show()
```

```python
sns.displot(data=df, x='size', kind='ecdf', height=5, aspect=2)
plt.grid(which='major')
plt.minorticks_on()
plt.show()
```

```python
sns.displot(data=df, x='size', kind='ecdf', height=5, aspect=2)
plt.grid(which='minor')
plt.minorticks_on()
plt.show()
```

Uylarning narxi uchun chizamiz

```python
sns.displot(data=df, x='price', kind='ecdf', height=5, aspect=2)
plt.grid(which='both')
plt.minorticks_on()
plt.show()
```
