```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/merc.csv")
df.head()
```

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x='year', y='price')
plt.show()
```

```python
df = df[df['year'] > 2000]
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(data=df, x='year', y='price')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='year', y='price', s=100)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='year', y='price', s=100, hue='fuelType')
plt.show()
```

```python
plt.figure(figsize=(10,5))
#sns.scatterplot(data=df, x='year', y='price', s=100, hue='fuelType')
sns.regplot(data=df, x='year', y='price')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='year', y='price', s=100, hue='fuelType', style='transmission')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='year', y='price', s=100, hue='fuelType', style='transmission', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10,5))
#sns.scatterplot(data=df, x='year', y='price', s=100, hue='fuelType', style='transmission', alpha=0.5)
sns.swarmplot(data=df, x='year', y='price')
plt.show()
```

```python
plt.figure(figsize=(10,5))
#sns.scatterplot(data=df, x='year', y='price', s=100, hue='fuelType', style='transmission', alpha=0.5)
sns.swarmplot(data=df, x=df['year'], y=df['price'])
plt.xticks(rotation=90)
plt.show()
```
